```python
import numpy as np
from qiskit import QuantumCircuit, QuantumRegister, ClassicalRegister, transpile
from qiskit_aer import AerSimulator


DNA_MAPPING = {
    "A": "00",
    "C": "01",
    "G": "10",
    "T": "11"
}

def encodeDna(s):
  s = s.upper()
  encoded = ""
  for char in s:
    if char in DNA_MAPPING:
      encoded += DNA_MAPPING[char]
    else:
      raise ValueError(f"Carattere non valido: {char}")
  return encoded


DNA_T = "ACGTACGT" # sequenza da analizzare
DNA_P = "TA" # pattern da trovare

T = encodeDna(DNA_T)
P = encodeDna(DNA_P)

# L'algoritmo funziona solo se |T| è potenza di 2
if not (len(T) & (len(T) - 1) == 0):
  raise ValueError("Errore: |T| deve essere una potenza di 2")

N = len(T)
M = len(P)
N_bases = N // 2
M_bases = M // 2
n = int(np.ceil(np.log2(N_bases)))
# Servono N/2 ancilla qubits per il fan-out, M-1 per andTree, più 1 dedicato per il phase kickback
a = max(N // 2, M - 1) + 1



def fanout():
  fo = QuantumCircuit(1 + N // 2, name="fanout") # controllo + ancilla qubits
  fo.cx(0, 1)
  copied = 1
  while copied < (N // 2):
    for i in range(copied):
      if (i + copied) < (N // 2):
        fo.cx(1 + i, copied + 1 + i)
    copied *= 2
  return fo.to_gate()

def cyclicShift():
  cs = QuantumCircuit(n + N + a, name="cyclicShift")
  for z in range(n):
    k = 2*(2**z) # *2 perchè ogni base ha 2 bit
    cs.append(fanout(), [z] + list(range(n + N, n + N + (N // 2)))) # copia l'indice in reg_A
    # i = strati paralleli, j = blocchi, x = elementi dei blocchi (prima metà)
    for i in range(int(np.log2(N) - np.log2(k)), 0, -1): # ordine inverso perché leftward
      anc_id = 0
      for j in range(N // ((2**i) * k)):
        for x in range(j * k * (2**i), (j * (2**i) + 1) * k):
          cs.cswap(n + N + anc_id, n + x, n + x + (2**(i - 1)) * k)
          anc_id += 1
    cs.append(fanout().inverse(), [z] + list(range(n + N, n + N + (N // 2)))) # ripulisce reg_A per il giro dopo
  return cs.to_gate()

def computation():
  comp = QuantumCircuit(n + N + M + a, name="computation")
  # list(range(n+N)) per K e T, list(range(n + N + M, n + N + M + a)) per A
  comp.append(cyclicShift(), list(range(n + N)) + list(range(n + N + M, n + N + M + a)))
  for i in range(M):
    comp.cx(n + i, n + N + i) # il CNOT esegue uno XOR, il risultato va nel secondo parametro
  return comp.to_gate()

def andTree():
  at = QuantumCircuit((M * 2) - 1, name="andTree")
  queue = list(range(M))
  for result in range(M - 1): # result è l'indice di A che scorriamo
    q1 = queue.pop(0)
    q2 = queue.pop(0)
    target = M + result
    at.ccx(q1, q2, target)
    queue.append(target) # aggiungiamo l'ancilla appena calcolato in fondo
  return at.to_gate()

def oracle():
  orc = QuantumCircuit(M * 2, name="oracle")
  orc.x(range(M)) # inverte P
  orc.append(andTree(), range((M * 2) - 1)) # combina i qubit di P in un qubit a
  orc.cx((M * 2) - 2, (M * 2) - 1) # phase kickback tra penultimo e ultimo a
  # uncomputation
  orc.append(andTree().inverse(), range((M * 2) - 1))
  orc.x(range(M))
  return orc.to_gate()

def diffusion():
  diff = QuantumCircuit(n, name="diffusion")
  diff.h(range(n))
  diff.x(range(n))
  diff.h(n - 1)
  diff.mcx(list(range(n - 1)), n - 1)
  diff.h(n - 1)
  diff.x(range(n))
  diff.h(range(n))
  return diff.to_gate()

# Costruiamo i gate una sola volta per velocizzare i tentativi del BBHT
comp_gate = computation()
comp_inv_gate = comp_gate.inverse()
oracle_gate = oracle()
diff_gate = diffusion()
simulator = AerSimulator(method="matrix_product_state") # Aer normale ha max 29 qubit

def grover(j):
  reg_K = QuantumRegister(n, name="k") # registro indice
  reg_T = QuantumRegister(N, name="t") # registro stringa
  reg_P = QuantumRegister(M, name="p") # registro pattern
  reg_A = QuantumRegister(a, name="a") # registro ancilla
  cr = ClassicalRegister(n, name="c")

  qc = QuantumCircuit(reg_K, reg_T, reg_P, reg_A, cr)

  # Popoliamo i registri usando identità e bit-flip come indicato nel paper
  for i in range(N):
    if T[i] == "1":
      qc.x(reg_T[i])
    elif T[i] == "0":
      qc.id(reg_T[i])
    else:
      raise ValueError("Errore nella codifica della stringa T")

  for i in range(M):
    if P[i] == "1":
      qc.x(reg_P[i])
    elif P[i] == "0":
      qc.id(reg_P[i])
    else:
      raise ValueError("Errore nella codifica del pattern")

  # Sovrapposizione reg_K
  qc.h(reg_K)

  # Ancilla |-> per il phase kickback
  qc.x(reg_A[a - 1])
  qc.h(reg_A[a - 1])

  for _ in range(j):
    qc.append(comp_gate, range(qc.num_qubits)) # shift + XOR
    qc.append(oracle_gate, reg_P[:] + reg_A[:M - 1] + [reg_A[a - 1]])
    qc.append(comp_inv_gate, range(qc.num_qubits)) # XOR + shift
    qc.append(diff_gate, reg_K[:])

  # Misura reg_K con shots=1
  qc.measure(reg_K, cr)
  qc_transpiled = transpile(qc, simulator)
  result = simulator.run(qc_transpiled, shots=1).result()
  counts = result.get_counts()

  # Converte la stringa binaria misurata in un intero e lo restituisce
  bitstring = list(counts.keys())[0]
  return int(bitstring, 2)

def verify(k):
  for i in range(M):
    if T[(2*k + i) % N] != P[i]:
      return False
  return True

def bbht():
  lam = 1.34
  m = 1.0
  max_steps = int(np.ceil(9 * np.sqrt(N_bases)))
  total_steps = 0
  attempt = 1

  while total_steps <= max_steps:
    # Estrai j casualmente in [0, floor(m) - 1]
    j = np.random.randint(0, max(1, int(np.floor(m))))

    # Esegui grover
    k = grover(j)

    # Aumenta passi totali
    total_steps += max(1, j)
    print(f"Tentativo {attempt} (m={m:.2f}, j={j}) -> indice misurato k={k}")

    if verify(k):
      return k
    else:
      m = min(lam * m, np.sqrt(N_bases)) # allarga la finestra
      attempt += 1

  print("Pattern non presente nella stringa.")
  return None

risultato = bbht()
if risultato is not None:
  print(f"Match confermato all'indice k = {risultato}!")
```
